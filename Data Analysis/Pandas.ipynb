### Pandas-DataFrame and Series
Pandas is a powerfull data manipulation library in python. Widely used for data analysis and data cleaning. It provides two primary data structures: Series and dataframe. A series is a one-dimentional array-like object, while a dataframe is a two-dimentional, size-mutable and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
# Series
data = [1, 2, 3, 4, 5]
series = pd.Series(data)
print("Series\n", series)
print(type(series))
```

```python
# Create a series from dictionary
data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10, 20, 30]
index = ['a', 'b', 'c']
pd.Series(data, index=index)
```

```python
# Dataframe
# Create a Dataframe from a dictionary of list

data = {
    'Name': ["Trupti", "Shweta", "Mayuri"],
    "Age": [21, 22, 18],
    "City": ["Pune", "Nagpur", "Gadchiroli"]
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Create a Datafrme from list of dictionaries

data = [
    {"Name": "Trupti", "Age": 21, "City": "Pune"},
    {"Name": "Shweta", "Age": 22, "City": "Nagpur"},
    {"Name": "Mayuri", "Age": 18, "City": "Gadchiroli"},
    {"Name": "John", "Age": 30, "City": "Florida"},
]

df = pd.DataFrame(data)
print(df)
```
