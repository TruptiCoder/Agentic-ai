#### File Operation - Read and write files
File handling is a crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary. This lesson will cover the basics of file handling, including reading and writing text files and binary files.

```python
## Read a whole file
with open("source.txt", 'r') as file:
    content = file.read()
    print(content)
```

```python
# Read line by line
with open("source.txt", 'r') as file:
    for line in file:
        print(line.strip()) # strip() removes new line
```

```python
# Writing a file (overriting)
with open("example.txt", 'w') as file:
    file.write("Hello world!\n")
    file.write("This is new line\n")
```

```python
# write a file without overriting
with open("example.txt", 'a') as file:
    file.write("\nThis line is appended")
```

```python
# writing list of lines
lines = ["First line", "Second line", "Third line"]
with open("example.txt", 'a') as file:
    file.writelines(lines)
```

```python
# Binary file
data = b'\x00\x01\x02\x03\x04'
with open("example.bin", 'wb') as file:
    file.write(data)
```

```python
with open("example.bin", 'rb') as file:
    content = file.read()
    print(content)
```

```python
# Read a text file and count the number of lines, words and characters.
with open("source.txt", 'r') as file:
    data = file.readlines()
    lines = len(data)
    words = sum(len(line.split()) for line in data)
    chars = sum(len(line) for line in data) 
print(f"lines {lines}, words {words}, characters {chars}")
```

w+ mode in python is used to open a file for both reading and writing. If  the file does not exist., it will be created. If the file exists, its content is truncated (i.e the file is overwritten)

```python
# Writing and then reading a file
with open("example.txt", "w+") as file:
    file.write("Hello world\n")
    file.write("This is a new line\n")

    # sets cursor to first line
    file.seek(0)

    # Read content
    content = file.read()
    print(content)
```

### working with file paths

```python
import os
new_dir = "my_dir"
os.mkdir(new_dir)
print(f"Directory '{new_dir}' is created")
```

```python
items = os.listdir('.')
print(items)
```

```python
# Joining paths
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd(), dir_name, file_name)
print(full_path)
```

```python
path = os.path.join(os.getcwd(), "example.txt")
if os.path.exists(path):
    print(f"The path {path} exists")
else: 
    print(f"Path {path} does not exists")
```
