### Loops

```python
# range(start, end, step)
for i in range(5, 1, -2):
    print("Hello", i)
```

```python
count = 0
while count < 3:
    print(count)
    count = count + 1
```

### List and List Comprehension
* Lists are ordered, mutable collections of items
* They can contain items of different datatypes

```python
list = [3, "Trupti", 56, 23.32, True]
type(list)
```

```python
print(list[1])
print(list[1:5:2]) # [start:end:step]
print(list[::])
print(list[:-1])
```

```python
fruits = ["Apple", "Banana", "Cherry", "Mango"]
for idx, fruit in enumerate(fruits): # enumerate gives index
    print("At index", idx, "Fruit =", fruit)
```

```python
# List comprehensions
list = []
for x in range(10):
    list.append(x**2)
print(list)
```

```python
[x**2 for x in range(10)] # same output as above
```

##### List Comprehension
- Basics Syntax [expression for item in iterable]
- with conditional logic [expression for item in iterable if condition]
- Nested List comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
even = [num**2 for num in range(10) if num%2 == 0]
print("Even", even)

#Nested
list1 = [1, 2]
list2 = ['a', 'b', 'c']
pair = [[i, j] for i in list1 for j in list2]
print("Pair", pair)
```

```python
prime = [num for num in range(2, 50) if all(num % i != 0 for i in range(2, int(num**0.5) + 1))]
prime
```

### Dictionaries
- Key value pairs
- Unordered
- Key must be uniqe and immutable

```python
dog = {"name": "Jupiter",
       "breed": "Golden retriver",
       "color": "Golden",
       "Girlfriend": "Goldy",
       "children": ["Gold", "Silver", "Iron"]}
print("---------Information of Dog----------")
for key, value in dog.items():
    print("---",key, "is", value)

# Can use dog.keys() and dog.values() as iterables
```

### Shallow copy
- Assignment (cat = dog): No copy is made. Both variables point to the same dictionary. Modifying either changes both.
- Shallow Copy (cat = dog.copy()): Creates a new dictionary container. Top-level changes in cat will not affect dog. 

```python
# This creates a reference
# cat = dog

#Shallow copy
cat = dog.copy()

cat["name"] = "nancy"
print("Cat", cat)
print("Dog", dog)
```

```python
# comprehensions
evens = {x:x**2 for x in range(10) if x%2 == 0}
print(evens)
```

```python
dict1 = {"a":1, "b":2}
dict2 = {"b":3, "c":4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
