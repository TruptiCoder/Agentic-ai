#### Exception handling in python

```python
try:
    a = b
    print("It did not run because of name error")
except:
    print("b is not assinged to Name error is there")
```

```python
try:
    a = 10 / 0
except ZeroDivisionError as err:
    print(f"There is error: {err}")
finally:
    print("This always prints")
```
