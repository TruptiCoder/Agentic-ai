### Functions in Python
- *args to take multiple arguments
- **kwargs to take multiple key value pairs

```python
def even_or_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
print(even_or_odd(20))
```

```python
def greet(name="Guest", *args, **kwargs):
    print(f"Hello {name} Welcome home!")
    print("I like following things:")
    for i in args:
        print(i, end=", ")
    print()
    print("My info is given below")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
        
greet("Trupti", "mango", "pineapple", "oranges", age=21, gender="Female")
```

#### Lambda function
Lambda functions are small anonymouse functions defined using the lambda keyword.
- lambda arguments: expression

```python
add = lambda a, b: a + b
add(4, 5)
```

##### map() function

```python
numbers = [1, 2, 3, 4, 5]
squred_nos = list(map(lambda x:x**2, numbers))
squred_nos
```

```python
nums1 = [1, 2, 3, 5, 6, 4]
nums2 = [4, 6, 3, 6, 3, 6]
added_nos = list(map(lambda x, y:x + y, nums1, nums2))
added_nos
```

```python
words = ["mango", "banana", "orange"]
upper_words = list(map(str.upper, words))
upper_words
```

#### Filter function

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8]
evens = list(filter(lambda num:num % 2 == 0 and num < 5, nums))
evens
```
