### Custom exception (Raise or Throw an exception)

```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year = int(input("Enter the year of birth: "))
age = 2026 - year

try:
    if age <= 30 and age >= 20:
        print("Age is valid, so you can apply for exam")
    else:
        raise dobException
except dobException:
    print("Sorry, your age is not withing the range")
```
