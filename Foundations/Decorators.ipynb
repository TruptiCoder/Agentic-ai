### Decorators
Decorators are a powerful and flexible feature in python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code.

```python
# Function copy
def welcome():
    return "Welcome to the python course"

welcome()
```

```python
wel = welcome
print(wel())
del welcome
print(wel())
```

```python
# closures - function inside function

def main_welcome(msg):
    def sub_welcome():
        print("welcome to the advance python course")
        print(msg)
        print("Please learn this concepts properly")
    return sub_welcome()
```

```python
main_welcome("Welcome to the course")
```

```python
def main_welcome(func):
    def sub_welcome():
        print("welcome to the advance python course")
        print(func("Welcome everyone!!!!!!"))
        print("Please learn this concepts properly")
    return sub_welcome()
```

```python
main_welcome(len)
```

```python
# Decorators
def main_welcome(func):
    def sub_welcome():
        print("welcome to the advance python course")
        func()
        print("Please learn this concepts properly")
    return sub_welcome()
```

```python
def course_introduction():
    print("This is an advance python course")

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("This is an advance python course")
```

```python
# Example
def my_decorator(func):
    def wrapper():
        print("Something is happing before function is called.")
        func()
        print("Something is happening after the function is called")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello")
```

```python
say_hello()
```

```python
# Decorators with arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def add(a, b):
    print(a + b)
```

```python
add(5, 7)
```
