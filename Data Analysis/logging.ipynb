### Python Logging
Logging is crucial aspect of any application, providing a way to track events, errors, and operational information. Python's built-in logging module offers a flexible framework for emmiting log messages from python programs.

```python
import logging

# Configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

# log messages
logging.debug("This is a debug message")
logging.info("This is info message")
logging.warning("This is warning")
logging.error("This is an error")
logging.critical("This is critical message")
```

##### Log levels
* Debug: Detailed info
* Info: Confirmation that things are working
* Warning: Somthing unexpected happend or some problem in near future but still working
* Error: Serious problem due to which software not working
* Critical: Very serious error, indication that the program itself may be unable to continue running.

```python
import logging

# Configurations
logging.basicConfig(
    filename='aap.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
)
```
